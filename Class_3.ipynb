### **2.1 What is a Series?**

### **3.2 Creating a DataFrame from a Dictionary**

```python
import pandas as pd
import numpy as np

survey = pd.DataFrame({
    "ID":[101,102,103,104,105,106],
    "Name":["Alice","Bob","Carol","David","Eva","Bob"],
    "Age":[22,np.nan,21,23,22,np.nan],
    "Department":["IIS","CSE","BBA","IIS","CSE","CSE"],
    "CGPA":[3.80,3.65,np.nan,3.75,3.88,3.65],
    "Gender":["Female","Male","Female","Male","Female","Male"]
})

print(survey)
```

### **8.1 Missing Values**

Missing values occur when information is unavailable.

```python
#Example 1: Detect Missing Values

print(survey.isnull())
```

```python
#Example 2: Count Missing Values

print(survey.isnull().sum())
```

### **8.2 notnull()**

Shows where data is present.



```python
print(survey.notnull())
```

```python
#Count available values.

print(survey.notnull().sum())
```

### **8.3 Filling Missing Values**

Suppose missing ages should be replaced with the average age.


```python
#Fill with Mean

survey = pd.DataFrame({
    "ID":[101,102,103,104,105,106],
    "Name":["Alice","Bob","Carol","David","Eva","Bob"],
    "Age":[22,np.nan,21,23,22,np.nan],
    "Department":["IIS","CSE","BBA","IIS","CSE","CSE"],
    "CGPA":[3.80,3.65,np.nan,3.75,3.88,3.65],
    "Gender":["Female","Male","Female","Male","Female","Male"]
})

survey["Age"] = survey["Age"].fillna(
    survey["Age"].mean()
)

print(survey)
```

```python
#Fill with Median

survey = pd.DataFrame({
    "ID":[101,102,103,104,105,106],
    "Name":["Alice","Bob","Carol","David","Eva","Bob"],
    "Age":[22,np.nan,21,23,22,np.nan],
    "Department":["IIS","CSE","BBA","IIS","CSE","CSE"],
    "CGPA":[3.80,3.65,np.nan,3.75,3.88,3.65],
    "Gender":["Female","Male","Female","Male","Female","Male"]
})

survey["CGPA"] = survey["CGPA"].fillna(
    survey["CGPA"].median()
)

print(survey)
```

```python
#Fill with Constant

survey = pd.DataFrame({
    "ID":[101,102,103,104,105,106],
    "Name":["Alice","Bob","Carol","David","Eva","Bob"],
    "Age":[22,np.nan,21,23,22,np.nan],
    "Department":["IIS","CSE","BBA","IIS","CSE","CSE"],
    "CGPA":[3.80,3.65,np.nan,3.75,3.88,3.65],
    "Gender":["Female","Male","Female","Male","Female","Male"]
})

survey["Department"] = survey["Department"].fillna("Unknown")

print(survey)
```

```python
#Fill with Forward Fill

survey = pd.DataFrame({
    "ID":[101,102,103,104,105,106],
    "Name":["Alice","Bob","Carol","David","Eva","Bob"],
    "Age":[22,np.nan,21,23,22,np.nan],
    "Department":["IIS","CSE","BBA","IIS","CSE","CSE"],
    "CGPA":[3.80,3.65,np.nan,3.75,3.88,3.65],
    "Gender":["Female","Male","Female","Male","Female","Male"]
})

survey["Age"] = survey["Age"].fillna(method="ffill")

print(survey)
```

```python
#Fill with Backward Fill

survey = pd.DataFrame({
    "ID":[101,102,103,104,105,106],
    "Name":["Alice","Bob","Carol","David","Eva","Bob"],
    "Age":[22,np.nan,21,23,22,np.nan],
    "Department":["IIS","CSE","BBA","IIS","CSE","CSE"],
    "CGPA":[3.80,3.65,np.nan,3.75,3.88,3.65],
    "Gender":["Female","Male","Female","Male","Female","Male"]
})

survey["Age"] = survey["Age"].fillna(method="bfill")

print(survey)
```

### **8.8 Change Data Types**

```python
import pandas as pd

survey = pd.read_csv("StudentSurvey.csv")

print("Original Shape")

print(survey.shape)

survey = survey.drop_duplicates()

survey["Age"] = survey["Age"].fillna(
survey["Age"].mean()
)

survey["CGPA"] = survey["CGPA"].fillna(
survey["CGPA"].median()
)

survey.rename(
columns={
"Dept":"Department"
},
inplace=True
)

survey["Gender"] = survey["Gender"].replace({
"M":"Male",
"F":"Female"
})

survey.to_csv(
"StudentSurvey_Cleaned.csv",
index=False
)

print("Cleaning Completed")
```

```python
#Mini Project 1: Student Survey Cleaning

import pandas as pd
import numpy as np

survey = pd.DataFrame({
    "ID":[1,2,3,4,4],
    "Name":["Alice","Bob","Carol","David","David"],
    "Age":[22,np.nan,21,23,23],
    "CGPA":[3.8,3.6,np.nan,3.9,3.9]
})

print("Original Dataset")
print(survey)

# Remove duplicates
survey = survey.drop_duplicates()

# Fill missing values
survey["Age"] = survey["Age"].fillna(survey["Age"].mean())
survey["CGPA"] = survey["CGPA"].fillna(survey["CGPA"].median())

# Rename columns
survey.rename(columns={"CGPA":"GPA"}, inplace=True)

# Convert Age to integer
survey["Age"] = survey["Age"].astype(int)

print("\nCleaned Dataset")
print(survey)
```
